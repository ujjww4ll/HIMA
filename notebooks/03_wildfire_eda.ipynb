# 03 — Wildfire Exploratory Data Analysis (NASA FIRMS 2021–2024)

Explores VIIRS S-NPP Collection 2 active-fire detections across Northeast India (2021–2024), validating `type == 0` vegetation fire filtering vs industrial flares (`type == 2`), monthly seasonality, FRP distributions, and hotspot maps.

```
import json, pandas as pd
from pathlib import Path
from IPython.display import Image, display

ROOT = Path('..')
f_meta = json.loads((ROOT / 'data/reports/firms_ingestion_metadata.json').read_text(encoding='utf-8'))
print('Years ingested:', f_meta['years_ingested'])
print('Type counts:', f_meta['type_value_counts'])
print('Confidence counts:', f_meta['confidence_value_counts'])
pd.Series(f_meta['detections_by_state'], name='detections')
```
