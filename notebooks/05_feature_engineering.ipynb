# 05 — Leakage-Safe Weather & Antecedent Rainfall Feature Engineering

Inspects ERA5-Land daily & 7/15-day rolling weather features (with missing 2023-08/09/11/12 preserved as `NaN`), separate IMD 0.25° daily & antecedent rainfall lags (`1/3/7/15` days), and feature correlations with wildfire presence.

```
import json, pandas as pd
from pathlib import Path
from IPython.display import Image, display

ROOT = Path('..')
w_meta = json.loads((ROOT / 'data/reports/weather_features_metadata.json').read_text(encoding='utf-8'))
print('ERA5 grid pts:', w_meta['used_era5_grid_points'], '| IMD grid pts:', w_meta['used_imd_grid_points'])
print('ERA5 missing rows flagged:', w_meta['era5_missing_rows_flagged'])
wf = pd.read_parquet(ROOT / 'data/processed/wildfire.parquet')
wf.groupby('split')['fire_detected'].agg(['count', 'sum', 'mean'])
```
